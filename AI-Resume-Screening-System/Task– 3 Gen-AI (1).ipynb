## Step 1: Setup Environment

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableLambda

print("Libraries Imported")
```

## Step 2: Define Prompt Templates

```python
extract_prompt = PromptTemplate(
    input_variables=["resume"],
    template="""
Extract the following details:
- Skills
- Experience
- Tools

Resume:
{resume}
"""
)

match_prompt = PromptTemplate(
    input_variables=["resume", "job_desc"],
    template="""
Compare resume with job description.

Resume:
{resume}

Job Description:
{job_desc}

Return matched and missing skills.
"""
)

score_prompt = PromptTemplate(
    input_variables=["data"],
    template="""
Give a score between 0-100 based on skill match.
Also provide a short explanation.

{data}
"""
)
```

## Step 3: Core Logic

```python
def evaluate_resume(resume, job_desc):
    resume_list = resume.lower().split(", ")
    job_list = job_desc.lower().split(", ")

    matched = [skill for skill in resume_list if skill in job_list]
    missing = [skill for skill in job_list if skill not in resume_list]

    score = int((len(matched) / len(job_list)) * 100)

    explanation = f"""
Matched Skills: {matched}
Missing Skills: {missing}
Score calculated based on matching skills.
"""

    return score, explanation
```

## Step 4: Build Pipeline

```python
def resume_pipeline(resume):
    score, explanation = evaluate_resume(resume, job_description)

    return f"""
Score: {score}/100

Explanation:
{explanation}
"""

chain = RunnableLambda(resume_pipeline)
```

## Step 5: Input Data

```python
job_description = "python, machine learning, nlp, sql"

strong_resume = "python, machine learning, nlp, sql, 3 years experience"
average_resume = "python, sql, 1 year experience"
weak_resume = "basic computer knowledge"
```

## Step 6: Run the Model

```python
print("=== STRONG RESUME ===")
print(chain.invoke(strong_resume))

print("\n=== AVERAGE RESUME ===")
print(chain.invoke(average_resume))

print("\n=== WEAK RESUME ===")
print(chain.invoke(weak_resume))
```
